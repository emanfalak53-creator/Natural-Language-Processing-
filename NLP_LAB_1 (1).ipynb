```python
# Task 06: String Cleaning Methods


# Part (a)
print("----- Part (a) -----")

text = "   MACHINE Learning is AMAZING!!!   "

text = text.strip()
print("After strip():", text)

text = text.lower()
print("After lower():", text)

text = text.replace("!", "")
print("After replace():", text)


# Part (b)
print("\n----- Part (b) -----")

words = text.split()
print("Words:", words)

uppercase_words = [word.upper() for word in words]
print("Uppercase words:", uppercase_words)

joined_text = " | ".join(uppercase_words)
print("Joined text:", joined_text)


# Part (c)
print("\n----- Part (c) -----")

sentence = "I love Java and Java is popular"

updated_sentence = sentence.replace("Java", "Python")

print("Updated sentence:", updated_sentence)

position = updated_sentence.find("Python")

print("Starting position of first 'Python':", position)


# Part (d)
print("\n----- Part (d) -----")

def normalize(text):
    text = text.strip()
    text = text.lower()
    text = text.replace(",", "")
    text = text.replace(".", "")

    return text


text1 = "  Hello, World.  "
text2 = "  Natural Language Processing, is interesting.  "

print("Original:", text1)
print("Cleaned:", normalize(text1))

print("Original:", text2)
print("Cleaned:", normalize(text2))
```

```python
# Task 07: String Validation Methods


# Part (a)
print("----- Part (a) -----")

text = "NLP is the future of AI research"

starts_with_nlp = text.startswith("NLP")
ends_with_research = text.endswith("research")
future_position = text.find("future")

print("Does the text start with 'NLP'?", starts_with_nlp)
print("Does the text end with 'research'?", ends_with_research)
print("Starting position of 'future':", future_position)


# Part (b)
print("\n----- Part (b) -----")

tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(token, "-> Alphabetic")
    elif token.isdigit():
        print(token, "-> Numeric")
    else:
        print(token, "-> Mixed")


# Part (c)
print("\n----- Part (c) -----")

words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)


# Part (d)
print("\n----- Part (d) -----")

def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"


tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in tokens:
    label = validate_token(token)
    print(token, "->", label)
```

```python
# Task 08: Functions and Lambda for NLP Pipelines


# Part (a)
# Regular function
def clean(text):
    text = text.strip()
    text = text.lower()
    return text


# Equivalent lambda function
clean_lambda = lambda text: text.strip().lower()


# Test the functions
text = "  HELLO Natural Language Processing  "

print("Regular function:", clean(text))
print("Lambda function:", clean_lambda(text))


# Part (b)
# Function to tokenize the text
def tokenize(text):
    return text.split()


# Function to remove words shorter than or equal to n
def remove_short(tokens, n):
    result = []

    for word in tokens:
        if len(word) > n:
            result.append(word)

    return result


# Given sentence
sentence = "I love working with NLP tools in Python daily"

tokens = tokenize(sentence)
filtered_tokens = remove_short(tokens, 3)

print("\nTokens:", tokens)
print("After removing short words:", filtered_tokens)


# Part (c)
# Function to find the length of each word
def word_lengths(tokens):
    result = {}

    for word in tokens:
        result[word] = len(word)

    return result


tokens = ["language", "model", "ai", "processing", "deep"]

lengths = word_lengths(tokens)

print("\nWord lengths:", lengths)


# Part (d)
# Complete NLP pipeline
def nlp_pipeline(text, min_len):
    # Clean the text
    text = clean(text)

    # Tokenize the text
    tokens = tokenize(text)

    # Remove short words
    tokens = remove_short(tokens, min_len)

    # Find the length of each remaining word
    result = word_lengths(tokens)

    return result


# First sentence
sentence1 = "  Natural Language Processing is very interesting  "

# Second sentence
sentence2 = "Artificial Intelligence helps computers learn"


print("\nPipeline result for sentence 1:")
print(nlp_pipeline(sentence1, 3))

print("\nPipeline result for sentence 2:")
print(nlp_pipeline(sentence2, 3))
```
