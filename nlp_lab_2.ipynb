```python
import PyPDF2
import csv

pdf_file = "assignment 2.pdf"

reader = PyPDF2.PdfReader(pdf_file)
pages_data = []

for i, page in enumerate(reader.pages, start=1):
    raw_text = page.extract_text() or ""
    words = raw_text.split()
    word_count = len(words)
    char_count = len(raw_text.replace(" ", ""))

    data = {
        "page_number": i,
        "raw_text": raw_text,
        "word_count": word_count,
        "char_count": char_count
    }

    pages_data.append(data)

print("First 2 entries:")
print(pages_data[:2])

non_zero_pages = [p for p in pages_data if p["word_count"] > 0]

if non_zero_pages:
    highest = max(non_zero_pages, key=lambda x: x["word_count"])
    lowest = min(non_zero_pages, key=lambda x: x["word_count"])

    print("\nPage with highest word count:")
    print("Page:", highest["page_number"])
    print("Word count:", highest["word_count"])
    print(highest["raw_text"])

    print("\nPage with lowest non-zero word count:")
    print("Page:", lowest["page_number"])
    print("Word count:", lowest["word_count"])

total_words = sum(p["word_count"] for p in pages_data)
average = total_words / len(pages_data)

print("\nAverage word count:", round(average, 2))

class PhraseNotFoundError(Exception):
    pass

phrase = input("\nEnter search phrase: ").lower()

try:
    found = False

    for page in pages_data:
        count = page["raw_text"].lower().count(phrase)

        if count > 0:
            print(
                "Page:",
                page["page_number"],
                "Occurrences:",
                count
            )
            found = True

    if not found:
        raise PhraseNotFoundError("Phrase not found in any page.")

except PhraseNotFoundError as e:
    print(e)

with open("pdf_summary.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["page_number", "word_count", "char_count"])

    for page in pages_data:
        writer.writerow([
            page["page_number"],
            page["word_count"],
            page["char_count"]
        ])

print("\nCSV contents:")

with open("pdf_summary.csv", "r", encoding="utf-8") as file:
    reader_csv = csv.reader(file)

    for row in reader_csv:
        print(row)
```

```python
import PyPDF2
import re
import nltk

from nltk.corpus import stopwords

nltk.download("stopwords")

pdf_file = "assignment 2.pdf"

reader = PyPDF2.PdfReader(pdf_file)

text = ""

for page in reader.pages:
    text += page.extract_text() or ""

print("Characters before cleaning:", len(text))

text = re.sub(r"https?://\S+|www\.\S+", "", text)

text = re.sub(r"\d+", "", text)

text = re.sub(r"[^a-zA-Z\s.]", "", text)

text = re.sub(r"\s+", " ", text)

print("Characters after cleaning:", len(text))

text = text.lower()

sentences = re.split(r"[.!?]", text)

print("Sentence count before removing empty:", len(sentences))

sentences = [sentence.strip() for sentence in sentences]

sentences = [sentence for sentence in sentences if sentence]

print("Sentence count after removing empty:", len(sentences))

stop_words = set(stopwords.words("english"))

final_sentences = []

for sentence in sentences:
    words = sentence.split()

    words = [
        word for word in words
        if word not in stop_words
    ]

    if len(words) >= 5:
        final_sentences.append(" ".join(words))

print("Sentence count before filter:", len(sentences))
print("Sentence count after filter:", len(final_sentences))

print("\n3 Sample sentences:")

for sentence in final_sentences[:3]:
    print(sentence)

with open("nlp_ready_output.txt", "w", encoding="utf-8") as file:
    for sentence in final_sentences:
        file.write(sentence + "\n")

with open("nlp_ready_output.txt", "r", encoding="utf-8") as file:
    lines = [line.strip() for line in file.readlines()]

print("\nTotal lines:", len(lines))

if lines:
    longest = max(lines, key=len)
    shortest = min(lines, key=len)

    print("\nLongest sentence:")
    print(longest)

    print("\nShortest sentence:")
    print(shortest)
```

```python
from docx import Document
from collections import defaultdict
from pprint import pprint

doc = Document("sample.docx")

paragraphs = []

for paragraph in doc.paragraphs:
    paragraphs.append({
        "text": paragraph.text,
        "style": paragraph.style.name
    })

print("Total paragraphs:", len(paragraphs))

styles = set()

for paragraph in paragraphs:
    styles.add(paragraph["style"])

print("\nUnique styles:")
for style in styles:
    print(style)

headings = []
body = []
other = []

for paragraph in paragraphs:
    text = paragraph["text"]
    style = paragraph["style"]

    if style.startswith("Heading"):
        headings.append(paragraph)

    elif style == "Normal" or "Body" in style:
        body.append(paragraph)

    else:
        other.append(paragraph)

print("\nHeading count:", len(headings))
print("Body count:", len(body))
print("Other/Empty count:", len(other))

qualifying = []

for paragraph in body:
    words = paragraph["text"].split()

    if len(words) > 5:
        qualifying.append({
            "text": paragraph["text"],
            "word_count": len(words),
            "char_count": len(paragraph["text"].replace(" ", ""))
        })

if qualifying:
    highest = max(qualifying, key=lambda x: x["word_count"])
    lowest = min(qualifying, key=lambda x: x["word_count"])

    mean = sum(
        item["word_count"] for item in qualifying
    ) / len(qualifying)

    print("\nHighest word count:")
    print(highest)

    print("\nLowest word count above 5:")
    print(lowest)

    print("\nMean word count:", round(mean, 2))

result = defaultdict(lambda: {
    "count": 0,
    "texts": []
})

for paragraph in paragraphs:
    style = paragraph["style"]

    result[style]["count"] += 1
    result[style]["texts"].append(paragraph["text"])

result = dict(
    sorted(
        result.items(),
        key=lambda x: x[1]["count"],
        reverse=True
    )
)

for style in result:
    if len(result[style]["texts"]) > 2:
        result[style]["texts"] = result[style]["texts"][:2]

print("\nFinal dictionary:")
pprint(result)
```

```python
from docx import Document
import pandas as pd

from docx.oxml.ns import qn

doc = Document("sample.docx")

tables = []

for table in doc.tables:
    table_data = []

    for row in table.rows:
        row_data = []

        for cell in row.cells:
            row_data.append(cell.text)

        table_data.append(row_data)

    tables.append(table_data)

print("Total tables:", len(tables))

for i, table in enumerate(tables, start=1):
    rows = len(table)
    columns = len(table[0]) if table else 0

    print(
        "Table",
        i,
        "dimensions:",
        rows,
        "x",
        columns
    )

if tables:
    first_table = tables[0]

    headers = first_table[0]
    data = first_table[1:]

    df = pd.DataFrame(data, columns=headers)

    print("\nDataFrame:")
    print(df)

    print("\nEmpty cells per column:")

    for column in df.columns:
        count = df[column].isna().sum()

        count += (
            df[column]
            .astype(str)
            .str.strip()
            .eq("")
            .sum()
        )

        print(column, ":", count)

pre_table_paragraphs = []

for element in doc.element.body:
    if element.tag == qn("w:tbl"):
        break

    if element.tag == qn("w:p"):
        text = ""

        for paragraph in doc.paragraphs:
            if paragraph._p == element:
                text = paragraph.text
                break

        pre_table_paragraphs.append(text)

print("\nPre-table paragraph count:", len(pre_table_paragraphs))

with open(
    "docx_extracted.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write("--- PRE-TABLE PARAGRAPHS ---\n")

    for paragraph in pre_table_paragraphs:
        file.write(paragraph + "\n")

    file.write("\n--- TABLE DATA ---\n")

    for row in first_table:
        file.write(" | ".join(row) + "\n")

print("\nFile created successfully.")
```

```python
from docx import Document
import re
from collections import Counter

doc_a = Document("documentA.docx")
doc_b = Document("documentB.docx")

headings_a = []

for paragraph in doc_a.paragraphs:
    if paragraph.style.name.startswith("Heading"):
        headings_a.append(paragraph.text.strip())

headings_b = []

for paragraph in doc_b.paragraphs:
    if paragraph.style.name.startswith("Heading"):
        headings_b.append(paragraph.text.strip())

set_a = set(headings_a)
set_b = set(headings_b)

print("Common headings:")
print(set_a & set_b)

print("\nHeadings only in Document A:")
print(set_a - set_b)

print("\nHeadings only in Document B:")
print(set_b - set_a)

body_a = []

for paragraph in doc_a.paragraphs:
    if paragraph.style.name == "Normal":
        body_a.append(paragraph.text)

body_b = []

for paragraph in doc_b.paragraphs:
    if paragraph.style.name == "Normal":
        body_b.append(paragraph.text)

all_body = body_a + body_b

print("\nCount before deduplication:", len(all_body))

unique_body = []
seen = set()

for paragraph in all_body:
    clean = paragraph.strip().lower()

    if clean not in seen:
        seen.add(clean)
        unique_body.append(paragraph.strip())

print("Count after deduplication:", len(unique_body))

all_words = []

for paragraph in unique_body:
    text = re.sub(r"[^a-zA-Z\s]", "", paragraph.lower())
    words = text.split()
    all_words.extend(words)

stopwords = {
    "the", "is", "a", "an", "and",
    "or", "to", "of", "in", "on",
    "for", "with", "as", "by", "this"
}

filtered_words = []

for word in all_words:
    if word not in stopwords:
        filtered_words.append(word)

frequency = Counter(filtered_words)

print("\nTop 20 words:")

for word, count in frequency.most_common(20):
    print(word, ":", count)

word_counts = []

for paragraph in unique_body:
    word_counts.append(len(paragraph.split()))

if word_counts:
    average = sum(word_counts) / len(word_counts)
    longest = max(
        unique_body,
        key=lambda x: len(x.split())
    )
else:
    average = 0
    longest = ""

print("\nSummary:")
print("Total headings:", len(headings_a) + len(headings_b))
print("Unique body paragraphs:", len(unique_body))
print("Vocabulary size:", len(set(filtered_words)))
print("Average paragraph length:", round(average, 2))
print("Longest paragraph:")
print(longest)
```

```python
import requests

users_url = "https://jsonplaceholder.typicode.com/users"
posts_url = "https://jsonplaceholder.typicode.com/posts"

users_response = requests.get(users_url)
posts_response = requests.get(posts_url)

if users_response.status_code != 200:
    raise ConnectionError(
        "Users status code: " +
        str(users_response.status_code)
    )

if posts_response.status_code != 200:
    raise ConnectionError(
        "Posts status code: " +
        str(posts_response.status_code)
    )

users = users_response.json()
posts = posts_response.json()

if not isinstance(users, list):
    raise TypeError("Users data is not a list.")

if not isinstance(posts, list):
    raise TypeError("Posts data is not a list.")

joined = {}

for user in users:
    user_id = user["id"]

    user_posts = []

    for post in posts:
        if post["userId"] == user_id:
            user_posts.append(post["title"])

    joined[user["name"]] = {
        "email": user["email"],
        "post_count": len(user_posts),
        "titles": user_posts
    }

print("First 3 users:")

for name in list(joined.keys())[:3]:
    print(name, ":", joined[name])

most_posts_user = max(
    users,
    key=lambda user: sum(
        1 for post in posts
        if post["userId"] == user["id"]
    )
)

user_id = most_posts_user["id"]

user_posts = [
    post for post in posts
    if post["userId"] == user_id
]

print("\nUser with most posts:")
print("Name:", most_posts_user["name"])
print("Email:", most_posts_user["email"])
print("Company:", most_posts_user["company"]["name"])

print("Post titles:")

for i, post in enumerate(user_posts, start=1):
    print(i, ".", post["title"])

print("\nMatching users:")

for user in users:
    bs = user["company"]["bs"].lower()
    city = user["address"]["city"]

    if "synergies" in bs or len(city) > 8:

        count = sum(
            1 for post in posts
            if post["userId"] == user["id"]
        )

        print(
            user["name"],
            "|",
            city,
            "| Posts:",
            count
        )
```

```python
import requests

todos_url = "https://jsonplaceholder.typicode.com/todos"

response = requests.get(todos_url)

if response.status_code != 200:
    print("Error: Status code is", response.status_code)
    exit()

todos = response.json()

if not isinstance(todos, list):
    print("Error: Data is not a list.")
    exit()

if len(todos) == 0:
    print("Error: No todo records found.")
    exit()

print("Total todo records:", len(todos))

user_data = {}

for todo in todos:
    user_id = todo["userId"]

    if user_id not in user_data:
        user_data[user_id] = {
            "total": 0,
            "completed": 0,
            "incomplete": 0
        }

    user_data[user_id]["total"] += 1

    if todo["completed"]:
        user_data[user_id]["completed"] += 1
    else:
        user_data[user_id]["incomplete"] += 1

print("\nTodo summary:")

for user_id in user_data:
    data = user_data[user_id]

    check = (
        data["completed"] +
        data["incomplete"]
    ) == data["total"]

    print(
        user_id,
        data,
        "Check:",
        check
    )

rates = {}

for user_id, data in user_data.items():
    rate = (
        data["completed"] /
        data["total"]
    ) * 100

    rates[user_id] = round(rate, 1)

highest_user = max(rates, key=rates.get)
lowest_user = min(rates, key=rates.get)

average_rate = sum(rates.values()) / len(rates)

print("\nHighest completion rate:")
print(highest_user, ":", rates[highest_user], "%")

print("\nLowest completion rate:")
print(lowest_user, ":", rates[lowest_user], "%")

print("\nOverall average:")
print(round(average_rate, 1), "%")

users_response = requests.get(
    "https://jsonplaceholder.typicode.com/users"
)

users = users_response.json()

names = {}

for user in users:
    names[user["id"]] = user["name"]

sorted_users = sorted(
    rates.items(),
    key=lambda x: x[1],
    reverse=True
)

print("\nRank | Name | Total Tasks | Completed | Completion Rate %")

for rank, (user_id, rate) in enumerate(
    sorted_users,
    start=1
):
    total = user_data[user_id]["total"]
    completed = user_data[user_id]["completed"]

    print(
        f"{rank:<4} | "
        f"{names[user_id]:<20} | "
        f"{total:<11} | "
        f"{completed:<9} | "
        f"{rate:.1f}"
    )
```

```python
import requests
import pandas as pd
import math

users_url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(users_url)

if response.status_code != 200:
    raise ConnectionError(
        "Status code: " +
        str(response.status_code)
    )

users = response.json()

def flatten_user(user):
    return {
        "id": user["id"],
        "name": user["name"],
        "username": user["username"],
        "email": user["email"],
        "city": user["address"]["city"],
        "zipcode": user["address"]["zipcode"],
        "lat": user["address"]["geo"]["lat"],
        "lng": user["address"]["geo"]["lng"],
        "company_name": user["company"]["name"],
        "company_catchphrase": user["company"]["catchPhrase"]
    }

flattened_users = []

for user in users:
    flattened_users.append(
        flatten_user(user)
    )

print("Flattened users:")
print(flattened_users)

df = pd.DataFrame(flattened_users)

df["lat"] = df["lat"].astype(float)
df["lng"] = df["lng"].astype(float)

df["distance_from_origin"] = df.apply(
    lambda row: math.sqrt(
        row["lat"] ** 2 +
        row["lng"] ** 2
    ),
    axis=1
)

print("\nDataFrame:")
print(df.to_string())

greatest = df.loc[
    df["distance_from_origin"].idxmax()
]

print("\nUser with greatest distance:")
print(greatest)

northern = df[df["lat"] > 0]
southern = df[df["lat"] < 0]

print("\nNorthern Hemisphere:")
print(northern["name"].tolist())

print("\nSouthern Hemisphere:")
print(southern["name"].tolist())

north_average = northern[
    "distance_from_origin"
].mean()

south_average = southern[
    "distance_from_origin"
].mean()

print("\nNorthern average distance:", north_average)
print("Southern average distance:", south_average)

if north_average > south_average:
    print("Northern Hemisphere has higher average distance.")
else:
    print("Southern Hemisphere has higher average distance.")

posts_response = requests.get(
    "https://jsonplaceholder.typicode.com/posts"
)

comments_response = requests.get(
    "https://jsonplaceholder.typicode.com/comments"
)

if posts_response.status_code != 200:
    raise ConnectionError(
        "Posts status code: " +
        str(posts_response.status_code)
    )

if comments_response.status_code != 200:
    raise ConnectionError(
        "Comments status code: " +
        str(comments_response.status_code)
    )

posts = posts_response.json()
comments = comments_response.json()

comment_counts = {}

for comment in comments:
    post_id = comment["postId"]

    if post_id not in comment_counts:
        comment_counts[post_id] = 0

    comment_counts[post_id] += 1

result = []

for post in posts:
    post_id = post["id"]
    title = post["title"][:40]
    count = comment_counts.get(post_id, 0)

    result.append(
        (post_id, title, count)
    )

result.sort(
    key=lambda x: x[2],
    reverse=True
)

print("\nTop 5 most-commented posts:")

for item in result[:5]:
    print(item)

print("\nBottom 5 least-commented posts:")

for item in result[-5:]:
    print(item)
```
